## Concrete findings

1. Back-transformed log predictions estimate the **conditional median** rent, not the mean — luxury tails inflate RMSE more than MAE.
2. Errors are larger in outer districts with fewer listings (thin data).
3. Very large flats and rare amenity combinations dominate the worst absolute errors.
4. Ridge / BayesianRidge beat Dummy by a large margin on MAE.
5. Near-duplicate collapse + GroupKFold reduce leakage from re-posted listings.
